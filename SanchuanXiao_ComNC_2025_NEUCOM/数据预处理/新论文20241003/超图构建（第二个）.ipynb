# 现在根据行业划分，按照2012版证监会行业分类名称，Nnindnme

```python
# 使用 value_counts() 查看 Nnindnme 列中的唯一值及其对应的数量
nnindnme_counts = df0['Nnindnme'].value_counts()

# 显示结果
print(nnindnme_counts)
print("行业的数量",len(nnindnme_counts))
```

查看 Nnindnme 列，首先判断这一列有没有仅仅出现一次的值，如果有的话就删除，然后对于出现过多次的值，举个例子，假如这一列有个值为零售业，并且发现出现了4次，那么就把这4个公司的stkcd放入一个列表，

```python
# 计算 Nnindnme 列中每个值的出现次数
nnindnme_counts = df0['Nnindnme'].value_counts()

# 找到出现超过一次的值
repeated_nnindnme = nnindnme_counts[nnindnme_counts > 1].index

# 删除 Nnindnme 仅出现一次的行
df_filtered = df0[df0['Nnindnme'].isin(repeated_nnindnme)]

# 初始化一个字典来存储每个 Nnindnme 值对应的 Stkcd 列表
nnindnme_to_stkcd = {}

# 对于出现多次的 Nnindnme，将其对应的 Stkcd 放入列表
for nnindnme_value in repeated_nnindnme:
    stkcd_list = df_filtered[df_filtered['Nnindnme'] == nnindnme_value]['Stkcd'].tolist()
    nnindnme_to_stkcd[nnindnme_value] = stkcd_list

# 打印每个 Nnindnme 对应的 Stkcd 列表
for nnindnme_value, stkcds in nnindnme_to_stkcd.items():
    print(f"Nnindnme: {nnindnme_value}, Stkcds: {stkcds}")
```

```python
import pandas as pd
import numpy as np

# 假设 df0 是包含 Stkcd 和 Nnindnme 列的数据
# 我们已经有一个字典 nnindnme_to_stkcd，它存储了 Nnindnme 值和对应的 Stkcd 列表

# 163 个不同的 Stkcd（来自您之前的 df0）
stkcds = df0['Stkcd'].unique()

# 获取 y 列的值（即每个 Nnindnme 值的数量）
nnindnme_values = list(nnindnme_to_stkcd.keys())

# 初始化矩阵，形状为 (163, y)
matrix = np.zeros((len(stkcds), len(nnindnme_values)), dtype=int)

# 将 Stkcd 转换为索引映射
stkcd_to_idx = {stkcd: idx for idx, stkcd in enumerate(stkcds)}

# 填充矩阵
for col_idx, nnindnme_value in enumerate(nnindnme_values):
    stkcd_list = nnindnme_to_stkcd[nnindnme_value]
    for stkcd in stkcd_list:
        row_idx = stkcd_to_idx[stkcd]
        matrix[row_idx, col_idx] = 1

# 将矩阵转换为 DataFrame，行是 Stkcd，列是 Nnindnme
matrix_df = pd.DataFrame(matrix, index=stkcds, columns=nnindnme_values)

# 显示结果
matrix_df

# 如果需要保存结果为 CSV 文件
# output_path = r'data/work/matrix.csv'
# matrix_df.to_csv(output_path)
```
