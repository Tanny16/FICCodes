# 对于每一个股票stkcd，有很多天的数据，我想得到一张新表，新表记录每个股票在各自交易日中的平均日开盘价，然后构成的新表格第一列为股票id，第二列为平均日开盘价，并按平均日开盘价从大到小，给我代码就行

```python
import pandas as pd

# 计算每个股票的平均日开盘价
avg_open_price_by_stock = filtered_by_date.groupby('Stkcd')['日开盘价'].mean().reset_index()

# 重命名列名
avg_open_price_by_stock.columns = ['Stkcd', 'AvgOpenPrice']

# 按平均开盘价从大到小排序
avg_open_price_by_stock = avg_open_price_by_stock.sort_values(by='AvgOpenPrice', ascending=False)

# 输出新的表格
avg_open_price_by_stock
```

# rolling(window=5).mean()：使用滚动窗口计算涨跌幅的移动平均值。window 参数定义了滚动窗口的大小，例如 5 日、10 日、20 日和 30 日。

```python
import pandas as pd

# 假设 selected_columns 是包含 Stkcd 和 涨跌幅 的 DataFrame

# 定义函数来计算每只股票的移动平均线
def calculate_moving_averages(df):
    df['5日均线'] = df['涨跌幅'].rolling(window=5).mean().fillna(0)
    df['10日均线'] = df['涨跌幅'].rolling(window=10).mean().fillna(0)
    df['20日均线'] = df['涨跌幅'].rolling(window=20).mean().fillna(0)
    df['30日均线'] = df['涨跌幅'].rolling(window=30).mean().fillna(0)
    return df

# 按 Stkcd 分组并应用移动平均线计算
selected_columns_with_ma = selected_columns.groupby('Stkcd').apply(calculate_moving_averages)

# 显示结果
selected_columns_with_ma

# 如果需要将结果保存为 CSV 文件
# output_path = r'data/work/selected_columns_with_ma.csv'
# selected_columns_with_ma.to_csv(output_path, index=False)
# print(f"结果已保存为 {output_path}")
```

```python
import pandas as pd

# 假设 selected_columns 是您的 DataFrame

# 创建一个新的列 '涨跌幅分类'，如果 '涨跌幅' > 0 则为 1，否则为 0
selected_columns_with_ma['涨跌幅分类'] = selected_columns_with_ma['涨跌幅'].apply(lambda x: 1 if x > 0 else 0)

# 计算正和负的数量
positive_count = selected_columns_with_ma['涨跌幅分类'].sum()  # 涨跌幅为 1 的数量
negative_count = len(selected_columns_with_ma) - positive_count  # 涨跌幅为 0 的数量

# 计算正负数量的比例
ratio = positive_count / negative_count if negative_count != 0 else 'Inf'

# 输出结果
print(f"涨跌幅为正的数量: {positive_count}")
print(f"涨跌幅为负的数量: {negative_count}")
print(f"正负数量比例: {ratio}")

# 如果需要保存处理
```
