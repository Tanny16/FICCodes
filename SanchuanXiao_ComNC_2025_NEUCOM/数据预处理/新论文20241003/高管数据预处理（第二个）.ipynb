如果任意两只股票，存在相同的University，则筛选出来，并构造[a,b,c,d，e]列表，其中a代表第一只股票的id，b代表第二只股票的id，c代表University，d代表第一只股票筛选出来的University所对应的name，e代表第二只股票筛选出来的University所对应的name

```python
from itertools import combinations
# 按 University 分组，找出有相同 University 的记录
univ_groups = result_filtered.groupby('University')

# 初始化存储结果的列表
result_list = []



# 遍历每一个 University 组
for university, group in univ_groups:
    # 如果某个大学有多个不同的股票，找出组合
    if group['Stkcd'].nunique() > 1:
        # 获取这个组内所有股票、对应的人名和大学
        unique_stocks = group[['Stkcd', 'Name', 'University']].drop_duplicates()

        # 对不同的股票代码进行两两组合
        for (stkcd1, name1, _), (stkcd2, name2, _) in combinations(unique_stocks.values, 2):
            result_list.append([stkcd1, stkcd2, university, name1, name2])

# 将结果转换为 DataFrame 方便查看或保存
df_result = pd.DataFrame(result_list, columns=['StockA', 'StockB', 'University', 'NameA', 'NameB'])
```

# 对于这个表，要用到NameA NameB的信息，构造一个矩阵，矩阵的行和列均为，结合NameA NameB在对应位置填1，注意（在NameA NameB对应位置填1的同时，NameB NameA对应的位置也要填1，因此是个对称矩阵）

```python
import pandas as pd
import numpy as np

# 假设 'names' 是你已经给出的高管列表，'df_result' 是你的 DataFrame

# 创建一个 n x n 的零矩阵，n 是高管列表的长度
n = len(unique_names)
matrix2 = pd.DataFrame(np.zeros((n, n),dtype=int), index=unique_names, columns=unique_names)

# 遍历 df_result，将 NameA 和 NameB 对应的矩阵位置填为 1
for _, row in df_result.iterrows():
    name_a = row['NameA']
    name_b = row['NameB']
    
    if name_a in unique_names and name_b in unique_names:
        matrix2.at[name_a, name_b] = 1  # 填充 NameA -> NameB
        matrix2.at[name_b, name_a] = 1  # 填充 NameB -> NameA, 保证对称性

# 打印或保存结果矩阵
matrix2
```

```python
# 创建两个列表用于存储行和列的序号
row_indices = []
column_indices = []

# 遍历矩阵，检查每个元素是否为 1，如果是 1 则记录其行和列的序号
for row_idx, (row_name, row) in enumerate(matrix2.iterrows()):  # 使用 enumerate 获取行的序号
    for col_idx, (col_name, value) in enumerate(row.items()):  # 使用 enumerate 获取列的序号
        if value == 1:
            row_indices.append(row_idx+163 )  # 行序号
            column_indices.append(col_idx+163 )  # 列序号

# 将行序号和列序号组合成一个大列表
result = [row_indices, column_indices]

# 输出结果
result
```
