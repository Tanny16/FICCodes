# OLS

```python
end=pd.read_csv(r'OLS/1987.csv')
for i in range(1987,2017):
    tmp=pd.read_csv(rf'OLS/{str(i)}.csv')
    end=pd.concat([end,tmp],axis=0)
end=end.drop('Unnamed: 0',axis=1)
print('R^2: {:.6f}'.format(r2_score(end['true'].values,end['pred'].values)))
```
