```python
class LlmAgent:
    def __init__(self, api_key, messages=[]):
        openai.api_key = api_key
        self.messages = messages
        #self.browser = webdriver.Chrome()

    def run(self):
        response = openai.ChatCompletion.create(
            model='gpt-4',
            messages=self.messages,
            temperature=0
        )
        return response.choices[0].message['content']
    
    def environmental_feedback(self, feedback):
        self.messages.append({
            'role': 'system',
            'content': feedback
        })

    def continue_conversation(self, user_input):
        self.messages.append({
            'role': 'user',
            'content': user_input
        })

        assistant_response = self.run()
        self.messages.append({
            'role': 'assistant',
            'content': assistant_response
        })

        return assistant_response
```

```python
class CodeAgent:
    def __init__(self, api_key, model, messages=[]):
        openai.api_key = api_key
        self.model = model
        self.messages = messages
        
        #self.browser = webdriver.Chrome()

    def run(self):
        response = openai.ChatCompletion.create(
            model=self.model,   
            messages=self.messages,
            temperature=0
        )
        return response.choices[0].message['content']
    
    def environmental_feedback(self, feedback):
        self.messages.append({
            'role': 'system',
            'content': feedback
        })

    def continue_conversation(self, user_input):
        self.messages.append({
            'role': 'user',
            'content': user_input
        })

        assistant_response = self.run()
        self.messages.append({
            'role': 'assistant',
            'content': assistant_response
        })

        return assistant_response
```

```python
def element_selector(response):
    match = re.search(r"\[selector\]: \[(.*?)\]", response)

    if match:
        inner_text = match.group(1)
        items = re.findall(r"'(.*?)'", inner_text)
        return items
    else:
        print('error')
        
def related_html(keywords= [],page_info = str ,window =500):
    related_html = {}
    for i in keywords:
        index = page_info.find(i)
        related_html[i] = page_info[index-window:index+window]
    return related_html

def find_code(response):
    #matches = re.findall(r'(?<=\[code\]: )\s*([\w\s\.\,\(\)\'\"=\[\]]+)', response)
    #matches = re.findall(r'(?<=\[code\]: )\s*([\w\s\.\,\(\)\'\"=\[\]/]+)', response)
    #matches = re.findall(r'(?<=\[code\]: )\s*([\w\s\.\,\(\)\'\"=\[\]/]+?\))', response)
    matches = re.findall(r'\[code\]: (.*?)(?=\[code\]:|$)', response, re.DOTALL)
    code_list = []
    for match in matches:
        for line in match.split('\n'):
            clean_line = line.strip()
            if clean_line:
                code_list.append(clean_line)
    return code_list


def skill_query_from_task(response, n):
    # Extract task using regular expression
    task_match = re.search(r"'current task':\s*'([^']+)'", response)
    if not task_match:
        return "No task found in the response."

    else:
        
        task = task_match.group(1)
        skill_name = manager.retrieve_skill_scored(task,n_results=n)["ids"][0]
        feedback = manager.retrieve_skill_scored(task,n_results=n)["documents"][0]
        score = manager.retrieve_skill_scored(task,n_results=n)['distances'][0]
        return skill_name,feedback,score,task
```

```python
def code_trigger(code_list):
    ipython = get_ipython()
    
    for index, i in enumerate(code_list):
        try:
            result = ipython.run_cell(i)
            time.sleep(1.5)
            
            if result.error_in_exec:
                
                # Capturing the traceback in case of an error
                error_message = ''.join(traceback.format_exception(type(result.error_in_exec), 
                                                                    result.error_in_exec, 
                                                                    result.error_in_exec.__traceback__))
                return f"Error in cell[{index + 1}]:\n{error_message}"
        except Exception as e:
            # In case another exception is raised
            return f"Exception in cell[{index + 1}]: {str(e)} - {traceback.format_exc()}"
    
    return True
```

```python
def read_KG_from_file(filename):
    KG = []
    with open(filename, 'r') as file:
        for line in file:
            # Remove the square brackets and split by comma
            triplet = [item.strip() for item in line.strip().replace("[", "").replace("]", "").split(',')]
            KG.append(triplet)
    return KG

# This function call will read your KG.txt and store each line as a list in KG

def find_paths(entity, kgs, current_path=None):
    if current_path is None:
        current_path = []

    # Find matches where the head of the KG matches the current entity
    #matches = [kg for kg in kgs if kg[0] == entity]
    pattern = re.compile(r'\b' + re.escape(entity.lower()) + r'\b')

    matches = [kg for kg in kgs if pattern.search(kg[0].lower())]
    

    # If no matches found, just return the current path as a list of paths
    if not matches:
        return [current_path]

    paths = []
    for match in matches:
        # Create a new path extending from the current path
        new_path = current_path + [match]
        paths.extend(find_paths(match[2], kgs, new_path))

    return paths

def build_tree(paths, limit = 0):
    # limit: only longest
    # Find the longest length among the paths
    #max_length = max(len(path) for path in paths)
    #print(max_length)
    # Filter out the longest path(s)
    #longest_paths = [path for path in paths if len(path) == max_length - limit]
    #pprint.pprint(longest_paths)
    #tree = defaultdict(dict)
    tree = {}

    def add_path_to_tree(node, path):
        if not path:
            return
        src, rel, dest = path[0]
        if src not in node:
            node[src] = {}
        if rel not in node[src]:
            node[src][rel] = {}
        if dest not in node[src][rel]:
            node[src][rel][dest] = {}
        add_path_to_tree(node[src][rel][dest], path[1:])

    for path in paths:
    #for path in longest_paths:
        add_path_to_tree(tree, path)

    return tree

def get_paths_dict(entities_list):
    paths_dict = {}
    
    for entity in entities_list:
        path = find_paths(entity, KG)

        path_tree = build_tree(paths = path)
        paths_dict[entity] = path_tree
    return paths_dict
```
